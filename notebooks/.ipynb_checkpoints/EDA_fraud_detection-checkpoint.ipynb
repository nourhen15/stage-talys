# EDA — Détection de fraude bancaire

**Objectif :** explorer les données pour identifier les features réellement utiles à la détection de fraude.

Cette analyse s'inscrit dans la boucle circulaire du CRISP-DM : les conclusions ici peuvent remettre en question le
preprocessing ou la sélection de features utilisée dans le pipeline d'entraînement (Airflow).

**Dataset :** Credit Card Fraud Detection (ULB/Kaggle) — `initial_data.csv`


## 1. Chargement des données

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

# Charge le même dataset que celui utilisé par Airflow pour l'entraînement
df = pd.read_csv("../data/raw/initial_data.csv")
print(f"Dimensions : {df.shape[0]} lignes, {df.shape[1]} colonnes")
df.head()
```

## 2. Vue d'ensemble : qualité des données

```python
print("Valeurs manquantes par colonne :")
print(df.isnull().sum().sum(), "valeurs manquantes au total")

print(f"\nDoublons : {df.duplicated().sum()} lignes dupliquées")

df.describe()
```

## 3. Distribution de la variable cible (`Class`)

Point clé pour tout le projet : ce dataset est fortement déséquilibré (peu de fraudes). Ça justifie l'usage de
`class_weight='balanced'` et du F1-score plutôt que de l'accuracy dans le pipeline d'entraînement.

```python
counts = df['Class'].value_counts()
pct_fraude = counts[1] / counts.sum() * 100

fig = px.bar(
    x=['Normale', 'Fraude'], y=counts.values,
    labels={'x': 'Classe', 'y': 'Nombre de transactions'},
    title=f"Répartition des classes ({pct_fraude:.3f}% de fraude)",
    color=['Normale', 'Fraude'],
    color_discrete_map={'Normale': '#2E86AB', 'Fraude': '#E63946'}
)
fig.show()
```

## 4. Distribution de `Amount` et `Time`

`Amount` et `Time` sont les seules colonnes non transformées par PCA — ce sont celles où le feature engineering
manuel (log, extraction de l'heure) a le plus de sens.

```python
fig = px.histogram(
    df, x="Amount", color="Class", nbins=100, marginal="box",
    title="Distribution du montant (Amount) par classe",
    color_discrete_map={0: '#2E86AB', 1: '#E63946'},
    log_y=True
)
fig.show()

print("Statistiques Amount par classe :")
df.groupby('Class')['Amount'].describe()
```

```python
# Feature engineering exploratoire : heure de la journée à partir de Time (en secondes)
df['Hour'] = (df['Time'] // 3600) % 24

fraud_by_hour = df[df['Class'] == 1].groupby('Hour').size()
normal_by_hour = df[df['Class'] == 0].groupby('Hour').size()
taux_fraude_par_heure = (fraud_by_hour / (fraud_by_hour + normal_by_hour) * 100).fillna(0)

fig = px.line(
    x=taux_fraude_par_heure.index, y=taux_fraude_par_heure.values,
    labels={'x': 'Heure de la journée', 'y': '% de fraude'},
    title="Taux de fraude selon l'heure de la journée",
    markers=True
)
fig.show()
```

## 5. Corrélations des features avec la fraude

Les colonnes `V1`-`V28` sont issues d'une PCA (donc déjà décorrélées entre elles par construction), mais leur
corrélation avec `Class` varie beaucoup — certaines sont bien plus discriminantes que d'autres.

```python
correlations = df.drop(columns=['Hour']).corr()['Class'].drop('Class').sort_values()

fig = px.bar(
    x=correlations.values, y=correlations.index,
    orientation='h',
    title="Corrélation de chaque feature avec la classe (fraude=1)",
    labels={'x': 'Corrélation', 'y': 'Feature'},
    color=correlations.values,
    color_continuous_scale='RdBu_r'
)
fig.update_layout(height=800)
fig.show()
```

## 6. Importance des features selon le modèle déjà entraîné

On recharge le modèle XGBoost gagnant (exporté par Airflow) pour voir quelles features il utilise le plus —
un second point de vue, complémentaire à la corrélation brute ci-dessus.

```python
model = joblib.load("../data/model/model.pkl")

importances = pd.Series(model.feature_importances_, index=model.feature_names_in_).sort_values(ascending=False)

fig = px.bar(
    x=importances.values[:15], y=importances.index[:15],
    orientation='h',
    title="Top 15 des features les plus importantes (XGBoost)",
    labels={'x': 'Importance', 'y': 'Feature'}
)
fig.update_layout(yaxis={'categoryorder': 'total ascending'}, height=600)
fig.show()
```

## 7. Distribution des features les plus discriminantes, par classe

```python
top_features = importances.index[:6].tolist()

fig, axes = plt.subplots(2, 3, figsize=(16, 8))
for ax, feature in zip(axes.flatten(), top_features):
    sns.boxplot(data=df, x='Class', y=feature, ax=ax, palette=['#2E86AB', '#E63946'])
    ax.set_xticklabels(['Normale', 'Fraude'])
    ax.set_title(feature)
plt.tight_layout()
plt.show()
```

## 8. Conclusions — Features à retenir

**À compléter après avoir exécuté les cellules ci-dessus avec les vraies données**, mais la structure à suivre :

- Les features en tête de la section 6 (importance du modèle) et ayant une forte corrélation en section 5 sont les
  candidates les plus solides à conserver.
- Si `Hour` (section 4) montre un vrai pattern temporel, l'ajouter comme feature dans `preprocess_data` du DAG
  Airflow est justifié.
- Les features avec une importance quasi nulle en section 6 pourraient être candidates à un retrait (à tester dans
  la phase Experiments, en comparant les métriques avec/sans elles).

**Prochaine étape (boucle circulaire CRISP-DM) :** intégrer ces conclusions dans `preprocess_data` du DAG, puis
relancer l'entraînement pour vérifier l'impact sur les métriques (F1, AUC-ROC).
